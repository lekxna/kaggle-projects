```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

footballdata = pd.read_csv(
    "/kaggle/input/champions-league-matches-2025-2026/champions_league_matches.csv"
)

footballdata.describe()
footballdata = footballdata.dropna(axis=0)
footballdata.columns

y = footballdata.winner
features = ["home_saves","away_saves","home_shots_on_target"]

x = footballdata[features]
x.describe()
```

```python

le = LabelEncoder()

footballdata['home_team_num'] = le.fit_transform(footballdata['home_team'])
footballdata['away_team_num'] = le.fit_transform(footballdata['away_team'])

x = footballdata[['home_team_num', 'away_team_num']]
y = footballdata['result']  

model = DecisionTreeClassifier(random_state=1)
model.fit(x, y)
predictions = model.predict(x)
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=1
)

model.fit(x_train, y_train)
predictions = model.predict(x_test)

print("Accuracy:", accuracy_score(y_test, predictions))

```

```python
next_games = [
    ['Ajax', 'Arsenal'],
    ['Atalanta', 'Athletic Club'],
    ['Atlético Madrid', 'Barcelona'],
    ['Bayern Munich', 'Benfica'],
    ['Bodø/Glimt', 'Chelsea']
]

next_games_df = pd.DataFrame(next_games, columns=['home_team', 'away_team'])

next_games_df['home_team_num'] = le.transform(next_games_df['home_team'])
next_games_df['away_team_num'] = le.transform(next_games_df['away_team'])


x_next = next_games_df[['home_team_num', 'away_team_num']]


next_predictions = model.predict(x_next)


for game, pred in zip(next_games, next_predictions):
    home, away = game
    if pred == 'Home Win':
        print(f"{home} wins vs {away} (loser: {away})")
    elif pred == 'Away Win':
        print(f"{away} wins vs {home} (loser: {home})")
    else:  
        print(f"{home} vs {away}: Draw (no loser)")
```
